Dictionary

```python
d = {'name': 'Alice', 'age': 25}

keys = ['name', 'age']
values = ['Alice', 25]
d = dict(zip(keys, values))  # Output: {'name': 'Alice', 'age': 25}

d['gender'] = "male"

for key in d:
    print(key)

for key in d:
    print(d[key])


for value in d.values():
    print(value)
```

Recursion

```python
def factorial(n):
  if n == 0 or n==1:
    return 1
  return n * factorial(n-1)

n = int(input("enter the number"))
print(factorial(n))
```

```python
def fibo(n):
  if n == 0 or n == 1:
    return 1;
  num = fibo(n-1)+fibo(n-2)
  return num

def print_fibo(n):
  for i in range(0,n):
    print(fibo(i))

n = int(input("enter the number"))
print(print_fibo(n))
```
